## Create the meta catalog

```python
from astropy.table import Table, vstack
import h5py
```

```python
#read catalogs as astropy Table 
ACT_DR6 = Table.read("./cluster_catalogs/DR6_cluster-legacy-catalog_v1.0.fits", format='fits')
HSC_Camira = Table.read("./cluster_catalogs/camira_s23b_wide_sm.tbl", format='ascii', 
                        names=["ID","Name","RA","Dec","z_cl","N_mem","z_spec_bcg"])
DES_WAZP = Table.read("./cluster_catalogs/y6a2_dnf_wazp_v5.0.12.6801_clusters.fits", format='fits')
with h5py.File("./cluster_catalogs/y3_redmapper_v6.4.22+2_release.h5", "r") as f:
    cluster_group = f["catalog/cluster"]
    table_data = {col: cluster_group[col][:] for col in cluster_group.keys()}
    DES_RM = Table(table_data)
```

```python
#create a standardized meta catalog, with only ra, dec, redhift, observable and catalog_provenance columns

sub_ACT_DR6 = Table({
    'ra': ACT_DR6['RADeg'],
    'dec': ACT_DR6['decDeg'],
    'redshift': ACT_DR6['redshift'],
    'observable': ACT_DR6['fixed_y_c'],
    'catalog_provenance': ["ACT_DR6"] * len(ACT_DR6)  
})

sub_HSC_Camira = Table({
    'ra': HSC_Camira['RA'],
    'dec': HSC_Camira['Dec'],
    'redshift': HSC_Camira['z_cl'],
    'observable': HSC_Camira['N_mem'],
    'catalog_provenance': ["HSC_Camira"] * len(HSC_Camira)  
})

sub_DES_WAZP = Table({
    'ra': DES_WAZP['RA'],
    'dec': DES_WAZP['DEC'],
    'redshift': DES_WAZP['ZPHOT'],
    'observable': DES_WAZP['NGALS'],
    'catalog_provenance': ["DES_Y6_WAZP"] * len(DES_WAZP)  
})

sub_DES_RM = Table({
    'ra': DES_RM['ra'],
    'dec': DES_RM['dec'],
    'redshift': DES_RM['z_lambda'],
    'observable': DES_RM['lambda_chisq'],
    'catalog_provenance': ["DES_Y3_RM"] * len(DES_RM)  
})


meta_catalog = vstack([sub_ACT_DR6, sub_HSC_Camira, sub_DES_WAZP, sub_DES_RM])
```

```python
#write an hdf5 file with all input catalogs and the standardized catalog 

output_hdf5 = "cluster_catalogs/cluster_KP_meta_catalog.hdf5"

ACT_DR6.write(output_hdf5, format='hdf5', path='catalogs/act_dr6', append=True, overwrite=True)
HSC_Camira.write(output_hdf5, format='hdf5', path='catalogs/hsc_camira', append=True, overwrite=True)
DES_WAZP.write(output_hdf5, format='hdf5', path='catalogs/des_y6_wazp', append=True, overwrite=True)
DES_RM.write(output_hdf5, format='hdf5', path='catalogs/des_y3_redmapper', append=True, overwrite=True)

meta_catalog.write(output_hdf5, format='hdf5', path='concatenated/standardized_meta_catalog', append=True, overwrite=True)
```

```python
#look at the structure of the file 
with h5py.File(output_hdf5, "r") as f:
    print("--- Contents of Unified HDF5 File ---")
    
    def print_structure(name, obj):
        print(f"Path: /{name}  -->  Type: {type(obj).__name__}")
        
    f.visititems(print_structure)
```
